```python
#Apartado 5
#Como observamos que para el caso de ind_var20_0 = 0 exiten outliers
subset_var15 = src_train[src_train['ind_var20_0'] == 0]['var15']
#Despues calculamos los cuartiles y el IQR
Q1 = subset_var15.quantile(0.25)
Q3 = subset_var15.quantile(0.75)
IQR = Q3 - Q1
#Establecemos los límites (vallas)
limite_superior = Q3 + 1.5 * IQR
limite_inferior = Q1 - 1.5 * IQR
#Obtenemos los outliers
outliers = subset_var15[(subset_var15 < limite_inferior) | (subset_var15 > limite_superior)]

print("Límite superior para outliers: ",limite_superior)
print("Límite inferior para outliers: ",limite_inferior)
print("Cantidad de outliers detectados:",len(outliers))
print("Valores de los outliers encontrados en var15:")
print(outliers.unique())
```

```python
#Apartado 6
#a)
mediana = src_train["var127"].median()
src_train['var127_log_mediana'] = np.log(src_train['var127']) / np.log(mediana)

#b)
media = src_train["var127"].mean()
std = src_train["var127"].std()

src_train["var127_norm"] = (src_train["var127"] - media) / std
```

```python
#Apartado 7
plt.figure()

sns.kdeplot(src_train["var127"], label="Original", fill = True)
sns.kdeplot(src_train["var127_log_mediana"], label="Log base mediana", fill = True)
sns.kdeplot(src_train["var127_norm"], label="Normalizada", fill = True)

plt.title("Densidades de Var127 y transformaciones")
plt.legend()
plt.show()
```

```python
#Apartado 8
columnas_seleccionadas = [
    "ID", "var15", "num_var22_hace3", "num_var45_hace3", "num_var22_ult3", 
    "num_var22_hace2", "num_med_var45_ult3", "num_var45_ult3", "ind_var30_0", 
    "var38", "var3", "num_med_var22_ult3", "saldo_medio_var12_ult3", 
    "saldo_medio_var12_ult1", "saldo_var5", "saldo_var24", "saldo_var13", 
    "saldo_var12", "num_var24_0", "ind_var20_0", "num_var13", "var36", "TARGET",
    "var127_log_mediana", "var127_norm"
]

# Creamos el nuevo data.frame (en Python es un DataFrame de pandas)
vun_train = src_train[columnas_seleccionadas].copy()

# Verificamos las dimensiones y las primeras filas
print("Dimensiones de vun_train:", vun_train.shape)
print(vun_train.head())
```

```python
vun_train.dtypes
```

```python
#BLOQUE 4
#Apartado 1
from sklearn.model_selection import train_test_split

X = vun_train.drop(columns = ["ID", "TARGET"])
y = vun_train["TARGET"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state= SEED)
```

```python
#Apartado 2
#Como todas las variables son numericas y no hay valores nulos(no debemos imputar), pero al hacer regresion scalaremos las variables
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, f1_score
import statsmodels.api as sm

variables = [
    "var15",
    "var38",
    "var3",
    "saldo_var5",
    "saldo_var12",
    "num_var22_ult3",
    "ind_var30_0",
    "ind_var20_0",
    "var127_norm"
]

X_train_sel = X_train[variables]
X_test_sel = X_test[variables]

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_sel)
X_test_scaled = scaler.transform(X_test_sel)

X_train_sm = sm.add_constant(X_train_scaled)
X_test_sm = sm.add_constant(X_test_scaled)

# Entrenamos el modelo de regresión logística
modelo_sm = sm.Logit(y_train, X_train_sm).fit(maxiter=1000)

print("Resumen")
print(modelo_sm.summary())

# Predicciones probabilísticas y conversión a binarias (umbral 0.5)
modelo_log = LogisticRegression(max_iter= 1000) 
modelo_log.fit(X_train_scaled, y_train) 
pred_log = modelo_log.predict(X_test_scaled)

#El modelo logístico presenta un pseudo-R² del 5.18%, lo que indica una capacidad explicativa moderada, sin embargo una maytrica nos contradice,
#lo deducido en el apartado siguiente con el modelo nulo
#Vemos como alguna variable influye significativamente en el target, pero el modelo no converge completamente por lo que mpos indica problemas,
```

```python
#Apartado 3
#El modelo nulo siempre predice la clas emayoritaria
pred_nulo = np.zeros(len(y_test))

print("\n--- APARTADO 3: COMPARATIVA CON MODELO NULO ---")
print("Accuracy Modelo Logístico: ",accuracy_score(y_test, pred_log))
print("Accuracy Modelo Nulo:",      accuracy_score(y_test, pred_nulo))
print("F1-Score (Macro) Logístico", f1_score(y_test, pred_log, average='macro'))
print("F1-Score (Macro) Nulo:",     f1_score(y_test, pred_nulo, average='macro'))
#Vemos que nuestro modelo, predice como un modelo nulo es decir siempre la clase mayoritariamente, por lo que habria que buscar soluciones a
#este problema ya que no seria una buena prediccion
```

```python
#Aoartado 4

print("Regresión logística") 
print("Matriz de confusion")
print(confusion_matrix(y_test, pred_log)) 
print("Reporte de clasificacion")
print(classification_report(y_test, pred_log))

#A simple vista parece exitoso pero al fijarse en la matriz de consudion nos damos cuenta de qu eno, y que siempre predice la clase mayoritaria
#En casos de desbalanceo hay que fijarse en metricas como el F1 o el Roc-AUC
```

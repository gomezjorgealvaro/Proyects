```python
plt.figure(figsize=(10, 5))  # tamaño más ancho para que entren bien los dos gráficos

plt.subplot(1, 2, 1)
plt.hist(df['DEPARTURE_DELAY'].dropna(), bins=40, color='skyblue', edgecolor='black', label='Salida')
plt.title("Retrasos en salida")
plt.xlabel("Minutos")
plt.ylabel("Frecuencia")
plt.legend()

plt.subplot(1, 2, 2)  
plt.hist(df['ARRIVAL_DELAY'].dropna(), bins=40, color='lightcoral', edgecolor='black', label='Llegada')
plt.title("Retrasos en llegada")
plt.xlabel("Minutos")
plt.ylabel("Frecuencia")
plt.legend()

plt.tight_layout()

plt.savefig()
```

```python
# Leyenda personalizada con dos líneas
plt.figure(figsize=(8, 5))

plt.plot(df.groupby('MONTH')['DEPARTURE_DELAY'].mean(), label='Salida', color='blue', linestyle='--', marker='o')
plt.plot(df.groupby('MONTH')['ARRIVAL_DELAY'].mean(), label='Llegada', color='red', linestyle='-', marker='s')

plt.title("Retraso medio mensual")
plt.xlabel("Mes")
plt.ylabel("Minutos")

plt.legend(
    loc='upper left',
    title='Tipo de retraso',
    fontsize='medium',
    title_fontsize='large',
    shadow=True,
    frameon=True,
    facecolor='white',
    edgecolor='black'
)

plt.grid(True)
plt.xticks(range(1, 13))
plt.tight_layout()
plt.show()
```

# Visualización de mapas en Plotly

```python
import pandas as pd
import plotly.express as px


ciudades = pd.DataFrame({
    "Ciudad": ["New York", "Los Angeles", "Chicago", "Houston", "Miami"],
    "Latitud": [40.7128, 34.0522, 41.8781, 29.7604, 25.7617],
    "Longitud": [-74.0060, -118.2437, -87.6298, -95.3698, -80.1918],
    "Número de vuelos": [12000, 9500, 8000, 7000, 6500],
    "Costa": [True,True,False,True,True]
})

fig = px.scatter_geo(
    ciudades,
    lat="Latitud",
    lon="Longitud",
    text="Ciudad",
    color = "Costa",
    size="Número de vuelos",
    projection="natural earth",
    title="Ciudades con más vuelos"
)

fig.show()
```

```python
import pandas as pd
import plotly.graph_objects as go

rutas = pd.DataFrame({
    "origen": ["New York", "Los Angeles", "Chicago"],
    "destino": ["Miami", "Houston", "San Francisco"],
    "lat_origen": [40.7128, 34.0522, 41.8781],
    "lon_origen": [-74.0060, -118.2437, -87.6298],
    "lat_destino": [25.7617, 29.7604, 37.7749],
    "lon_destino": [-80.1918, -95.3698, -122.4194]
})

fig = go.Figure()

for i, row in rutas.iterrows():
    fig.add_trace(go.Scattergeo(
        locationmode='USA-states',
        lon=[row["lon_origen"], row["lon_destino"]],
        lat=[row["lat_origen"], row["lat_destino"]],
        mode='lines',
        line=dict(width=2, color='blue'),
        opacity=0.5
    ))

fig.add_trace(go.Scattergeo(
    lon=rutas["lon_origen"],
    lat=rutas["lat_origen"],
    mode='markers+text',
    text=rutas["origen"],
    textposition="top center",
    marker=dict(size=6, color='red'),
    name='Origen'
))

fig.add_trace(go.Scattergeo(
    lon=rutas["lon_destino"],
    lat=rutas["lat_destino"],
    mode='markers+text',
    text=rutas["destino"],
    textposition="bottom center",
    marker=dict(size=6, color='green'),
    name='Destino'
))

fig.update_layout(
    title_text="Rutas entre ciudades",
    showlegend=False
    
)

fig.show()
```

```python

estados = pd.DataFrame({
    "estado": ["CA", "TX", "NY", "FL", "IL"],
    "vuelos": [15000, 12000, 11000, 10000, 9000]
})

fig = px.choropleth(
    estados,
    locations="estado",
    locationmode="USA-states",
    color="vuelos",
    color_continuous_scale="Blues",
    scope="usa",
    labels={"vuelos": "Número de vuelos"},
    title="Número de vuelos por estado"
)

fig.show()
```
